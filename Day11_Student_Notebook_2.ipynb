### Part B — Raising Our Own ValueError

Python can raise a ValueError on its own — like when you convert `"twenty"` to `int`.

But **we** can also raise a ValueError deliberately — to signal that a value breaks our rules.

This is how validation functions work.

````python
# raise ValueError() — we trigger the error ourselves
# When a rule is broken, we raise a ValueError with a clear message

# Run this and observe the error

age = 150

if age > 99:
    raise ValueError("Age cannot be more than 99")

# The program stops at raise — that is intentional
# In a moment we will catch this with try/except
````

````python
# Now combine raise with try/except
# The validator RAISES the error. The caller CATCHES it.
# This is the pattern EVERY validator will follow.

# Run this and observe

def check_age(age):
    """Checks if age is between 16 and 99. Raises ValueError if not."""
    if age > 99:
        raise ValueError("Age cannot be more than 99")
    if age < 16:
        raise ValueError("Age must be at least 16")
    return True   # only reaches here if both checks passed

try:
    check_age(150)
    print("Age is valid!")     # only prints if no error was raised
except ValueError as e:        # 'e' holds our error message
    print(f"Validation failed: {e}")
````

### Part B — Building a Pattern One Piece at a Time

A regex pattern is a **description** — like giving directions.

Instead of saying *"a valid student ID"*, you write a pattern that says:
*"start, then M, then O, then R, then exactly 5 digits, then end."*

We will build the student ID pattern `r"^MOR\d{5}$"` from scratch — **one piece per cell**.

**Important:** Always write patterns with `r"..."` — the `r` means *raw string* and stops Python from interfering with the backslashes.

````python
# PIECE 1: Literals — match exact characters
# r"MOR" means: find the exact letters M, O, R in that order
#
# re.match(pattern, string) returns:
#   a match object  — if the pattern was found
#   None            — if the pattern was NOT found
#
# We use this like True/False in an if statement

import re

pattern = r"MOR"

# Run this and observe each result
print(re.match(pattern, "MOR12345"))   # starts with MOR
print(re.match(pattern, "STU12345"))   # does NOT start with MOR
print(re.match(pattern, "mor12345"))   # lowercase — is it the same?
````

````python
# PIECE 2: \d — matches any single digit (0 to 9)
# The \ is why we need r"..." — it is a special character in regex

pattern = r"\d"

# Run and observe
print(re.match(pattern, "5"))     # is 5 a digit?
print(re.match(pattern, "a"))     # is a a digit?
print(re.match(pattern, "1abc"))  # starts with a digit — partial match
````

````python
# PIECE 3: {n} — exactly n repetitions of the piece before it
# \d{5} means: exactly 5 digits in a row

pattern = r"\d{5}"

# Run and observe — which ones match and which do not?
print(re.match(pattern, "12345"))   # exactly 5 digits
print(re.match(pattern, "1234"))    # only 4 digits
print(re.match(pattern, "123456"))  # 6 digits — does this match? why?

# Notice: '123456' still matches because re.match() only checks the START
# We need anchors to fix this — that is the next piece
````

````python
# PIECE 4: ^ and $ — anchors
# ^ means: the pattern must start at the VERY BEGINNING of the string
# $ means: the pattern must end at the VERY END of the string
# Together: the ENTIRE string must match — nothing extra allowed

pattern = r"^\d{5}$"   # start, exactly 5 digits, end

# Run and observe
print(re.match(pattern, "12345"))    # exactly 5 digits
print(re.match(pattern, "123456"))   # 6 digits — does $ stop it now?
print(re.match(pattern, "1234"))     # only 4 digits
print(re.match(pattern, "1234a"))    # letter at the end

# Always use ^ and $ when validating — you want the WHOLE string to match
````

````python
# PIECE 5: Combine everything — the complete student ID pattern
#
# ^     = start of string
# MOR   = literal letters M, O, R
# \d{5} = exactly 5 digits
# $     = end of string
#
# Say it out loud before running:
# "START, then MOR, then exactly 5 digits, then END"

pattern = r"^MOR\d{5}$"

# Run and observe each result
print(re.match(pattern, "MOR12345"))    # valid student ID?
print(re.match(pattern, "MOR1234"))     # only 4 digits
print(re.match(pattern, "MOR123456"))   # 6 digits
print(re.match(pattern, "STU12345"))    # wrong prefix
print(re.match(pattern, "mor12345"))    # lowercase
````

### Part C — Two More Pattern Pieces We Will Need

````python
# PIECE 6: Character classes [ ] — match any ONE character from a set
# [A-Z]    = any uppercase letter
# [a-z]    = any lowercase letter
# [A-Za-z] = any letter, upper or lower
# [A-Za-z ] = any letter OR a space  (note the space inside)
# + after the bracket means: one or more of those characters

# Run and observe
print(re.match(r"^[A-Za-z ]+$", "Amina Odhiambo"))  # letters and space
print(re.match(r"^[A-Za-z ]+$", "John123"))          # has digits
print(re.match(r"^[A-Za-z ]+$", "Maria-Jose"))        # has a hyphen
````

````python
# PIECE 7: [17] — character class with specific characters
# [17] means: match EITHER 1 OR 7 — nothing else
#
# Phone pattern logic:
# ^0     = must start with 0
# [17]   = then either 7 or 1  → 07 or 01
# \d{8}  = then exactly 8 more digits
# $      = nothing else
# Total = 1 + 1 + 8 = 10 digits

pattern = r"^0[17]\d{8}$"

# Run and observe
print(re.match(pattern, "0712345678"))     # 07 — valid?
print(re.match(pattern, "0112345678"))     # 01 — valid?
print(re.match(pattern, "0812345678"))     # 08 — valid?
print(re.match(pattern, "071234567"))      # only 9 digits
print(re.match(pattern, "+254712345678")) # starts with +
````

### Part D — The Email Pattern (Most Complex — Read It Out Loud)

The email pattern looks complicated. But we can **read it piece by piece**.

An email looks like this: `amina@moringa.ac.ke`

| Part | What it matches | Pattern piece |
|------|----------------|---------------|
| `amina` | local part — word chars, dots, hyphens | `[\w\.-]+` |
| `@` | literal @ symbol | `@` |
| `moringa.ac` | domain — word chars, dots, hyphens | `[\w\.-]+` |
| `.` | literal dot (escaped!) | `\.` |
| `ke` | TLD — word characters | `\w+` |

Full pattern: `r"^[\w\.-]+@[\w\.-]+\.\w+$"`

Say it out loud: *START, local-part, @, domain, dot, TLD, END*

````python
# First: \w — word character (letter, digit, or underscore)
# \w+ means: one or more word characters

print(re.match(r"^\w+$", "amina"))      # letters are word chars
print(re.match(r"^\w+$", "amina123"))   # letters and digits
print(re.match(r"^\w+$", "amina!"))     # ! is NOT a word char
print(re.match(r"^\w+$", "amina.k"))    # dot is NOT a word char by itself
````

````python
# Next: \. — a LITERAL dot (escaped)
# In regex, a plain . means 'any character'
# To match an actual dot, we write \. to escape it

print(re.match(r"\.", "."))    # literal dot — matches
print(re.match(r"\.", "a"))    # 'a' is not a dot — does it match?
print(re.match(r".", "a"))     # unescaped dot — matches ANY character
````

````python
# Now the full email pattern
# Read it out loud before running:
# "START, local-part, @, domain, dot, TLD, END"

email_pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

test_emails = [
    "amina@moringa.ac.ke",
    "@.com",
    "notanemail",
    "amina@",
    "amina@moringa",
]

# Run and observe — which ones match? which do not?
for e in test_emails:
    result = re.match(email_pattern, e)
    print(f"'{e}' → {'MATCH' if result else 'NO MATCH'}")
````

### Combining Everything — register_student()

## Session Recap

| Concept | What it means | Example |
|---------|---------------|---------|
| Input Validation | Checking data meets rules before processing | Checking email format before saving |
| `raise ValueError` | Trigger an error with a message | `raise ValueError("Name is empty")` |
| `try/except` | Catch an error and handle it gracefully | Print a message instead of crashing |
| `import re` | Load the regex module | `import re` |
| `r"..."` | Raw string — always use for patterns | `r"^MOR\d{5}$"` |
| Literal | Match exact characters | `r"MOR"` |
| `\d` | Any single digit | `\d{5}` = exactly 5 digits |
| `[A-Za-z ]` | Any letter or space | `^[A-Za-z ]+$` |
| `{n}` | Exactly n repetitions | `\d{5}` |
| `{n,m}` | Between n and m repetitions | `[A-Za-z0-9]{6,10}` |
| `^` and `$` | Anchor to start and end | `^\d{5}$` |
| `re.match()` | Check if string matches pattern | Returns match object or None |
