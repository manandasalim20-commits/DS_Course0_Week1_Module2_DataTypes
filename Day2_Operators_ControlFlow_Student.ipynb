## Section 3 — Control Flow: Making Decisions

### 3.1 — The structure

Without control flow, every program runs top to bottom the same way every time — no choices, no branches.
Control flow lets your program ask: **"is this condition True?"** and take different paths.

```
# Pseudocode
IF condition is True:
    do this
ELSE IF another condition is True:
    do this instead
ELSE:
    nothing above matched — do this
```

In Python:
```python
if condition:
    # 4-space indent
elif another_condition:
    # 4-space indent
else:
    # 4-space indent
```

**Three rules:**
1. Every condition line ends with a colon `:`
2. Code inside each block is indented **exactly 4 spaces**
3. Python checks top to bottom and **stops at the first True condition**
