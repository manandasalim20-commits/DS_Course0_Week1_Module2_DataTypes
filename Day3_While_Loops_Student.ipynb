## Section 2 — The Counter Loop

Every while loop needs exactly **three parts** — without all three, either the loop never runs or it runs forever.

```
# 1. INITIALISE  — set up the variable before the loop starts
# 2. CONDITION   — checked at the top of every cycle (True = keep going, False = stop)
# 3. BODY + UPDATE — what runs each cycle; must change the condition variable
```

````python
# Follow along — the three parts are labelled

# Part 1 — Initialise
counter = 1

# Part 2 — Condition
while counter <= 5:

    # Part 3 — Body
    print("Count:", counter)

    # Part 3 also — Update (without this: infinite loop)
    counter += 1

print("Done. Final value:", counter)
````

## Section 3 — Input Validation Loop

A common real-world pattern: keep asking the user for input until they give something valid.
`if/elif/else` checks once and moves on. A while loop keeps asking until the answer is right.

````python
# Follow along — keep asking until the user enters a number between 0 and 100
# Run this cell. Try entering: -5, 150, hello, then a valid number

valid_input = False          # initialise — not valid yet

while not valid_input:       # keep looping while NOT valid
    raw = input("Enter a score (0–100): ")

    if raw.isdigit():        # check it's a number first
        score = int(raw)
        if 0 <= score <= 100:
            valid_input = True   # this breaks the loop on next check
        else:
            print("Out of range. Try again.")
    else:
        print("Not a number. Try again.")

print(f"\nValid score accepted: {score}")
````
