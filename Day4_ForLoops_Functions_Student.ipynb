### 2.2 — range()

```python
# Follow along: three ways to use range()

print("range(5):")
for i in range(5):
    print(i, end=" ")     # end=" " keeps output on one line

print("\nrange(1, 6):")
for i in range(1, 6):
    print(i, end=" ")

print("\nrange(0, 10, 2):")
for i in range(0, 10, 2):
    print(i, end=" ")
```

### 2.3 — Accumulator pattern

```python
# Follow along: total the scores using an accumulator
scores = [72, 85, 91, 60, 78]
total = 0                      # start at 0

for score in scores:
    total += score             # add each score to the running total

print(f"Total: {total}")
```

## Recap — What did we cover today?

| Concept | Syntax | When to use |
|---|---|---|
| For loop | `for item in iterable:` | Process every item in a collection |
| range() | `range(n)`, `range(a,b)`, `range(a,b,step)` | Loop a fixed number of times |
| Accumulator | `total = 0; total += x` | Build up a running sum/count |
| Define function | `def name(params):` | Write reusable logic once |
| Return | `return value` | Send a result back to the caller |
| Call function | `result = name(arg)` | Use the function wherever needed |
